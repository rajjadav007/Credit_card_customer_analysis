```python
outlier_column=[
    'Credit_Limit',
    'Total_Revolving_Bal',
    'Total_Trans_Amt'
]

cust[outlier_column].describe()
```

```python
plt.figure(figsize=(12, 5))

for i, col in enumerate(outlier_column, 1):
    plt.subplot(1, 3, i)
    sns.boxplot(y=cust[col])
    plt.title(col)

plt.tight_layout()
plt.show()
```

```python
outlier_summary = []

for col in outlier_column:
    Q1 = cust[col].quantile(0.25)
    Q3 = cust[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = cust[
        (cust[col] < lower_bound) |
        (cust[col] > upper_bound)
    ]

    outlier_summary.append({
        "Column": col,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Outlier Count": len(outliers),
        "Outlier Percentage": len(outliers) / len(cust) * 100
    })

outlier_summary = pd.DataFrame(outlier_summary)
display(outlier_summary)
```
