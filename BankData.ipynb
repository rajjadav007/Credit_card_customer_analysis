```python
numerical_columns = [
    "Customer_Age",
    "Dependent_count",
    "Months_on_book",
    "Total_Relationship_Count",
    "Months_Inactive_12_mon",
    "Contacts_Count_12_mon",
    "Credit_Limit",
    "Total_Revolving_Bal",
    "Avg_Open_To_Buy",
    "Total_Amt_Chng_Q4_Q1",
    "Total_Trans_Amt",
    "Total_Trans_Ct",
    "Total_Ct_Chng_Q4_Q1",
    "Avg_Utilization_Ratio"
]

display(cust[numerical_columns].dtypes)
```

```python
for col in numerical_columns:
    converted = pd.to_numeric(cust[col], errors="coerce")
    invalid = cust[col].notna() & converted.isna()

    print(f"{col}: {invalid.sum()} invalid values")
```
