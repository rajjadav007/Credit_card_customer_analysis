```python
outlier_summary = []

for col in outlier_column:
    Q1 = cust[col].quantile(0.25)
    Q3 = cust[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = cust[
        (cust[col] < lower_bound) |
        (cust[col] > upper_bound)
    ]

    outlier_summary.append({
        "Column": col,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Outlier Count": len(outliers),
        "Outlier Percentage": len(outliers) / len(cust) * 100
    })

outlier_summary = pd.DataFrame(outlier_summary)
display(outlier_summary)
```

# Customer Attrition Prediction Using Machine Learning

## Objective

The objective is to build a machine learning model that predicts whether a credit card customer is likely to leave the bank based on customer demographics, credit information, and transaction behavior.

## Approach

1. Prepare the features and target variable.
2. Remove customer identifiers and target-derived columns.
3. Split the data into training and testing sets.
4. Build preprocessing pipelines.
5. Train and compare classification models.
6. Evaluate accuracy, precision, recall, F1-score, and ROC-AUC.
7. Tune the best model and save the final pipeline.

```python
leakage_columns = [
    col for col in analysis_df.columns
    if col.startswith("Naive_Bayes_Classifier_")
]

# Remove customer ID and leakage columns
ml_df = analysis_df.drop(
    columns=["CLIENTNUM"] + leakage_columns,
    errors="ignore"
).copy()

# Separate features and target
X = ml_df.drop(columns=["Attrition_Flag"])

y = ml_df["Attrition_Flag"].map({
    "Existing Customer": 0,
    "Attrited Customer": 1
})

# Validate target labels
if y.isna().any():
    raise ValueError("Unexpected values found in Attrition_Flag")

y = y.astype(int)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer
```

```python
numeric_features = X_train.select_dtypes(
    include=["number"]
).columns.tolist()

categorical_features = X_train.select_dtypes(
    exclude=["number"]
).columns.tolist()\

numeric_features

categorical_features
```

```python
numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

numeric_pipeline
```

```python
categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

categorical_pipeline
```

```python
preprocessor = ColumnTransformer([
    ("num", numeric_pipeline, numeric_features),
    ("cat", categorical_pipeline, categorical_features)
])

preprocessor
```

```python
from sklearn.linear_model import LogisticRegression

logistic_model = Pipeline([
    ("preprocessor", preprocessor),
    ("model", LogisticRegression(
        max_iter=2000,
        class_weight="balanced",
        random_state=42
    ))
])

logistic_model.fit(X_train, y_train)
```

```python
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    roc_auc_score
)

y_pred = logistic_model.predict(X_test)
y_prob = logistic_model.predict_proba(X_test)[:, 1]

print("Accuracy:", round(accuracy_score(y_test, y_pred), 4))
print("ROC-AUC:", round(roc_auc_score(y_test, y_prob), 4))

print("\nClassification Report:")
print(classification_report(
    y_test,
    y_pred,
    target_names=["Existing Customer", "Attrited Customer"]
))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```

```python
from sklearn.ensemble import RandomForestClassifier, ExtraTreesClassifier
from sklearn.model_selection import StratifiedKFold, cross_validate

models = {
    "Logistic Regression": LogisticRegression(
        max_iter=2000,
        class_weight="balanced",
        random_state=42
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=300,
        min_samples_leaf=2,
        class_weight="balanced",
        random_state=42,
        n_jobs=1
    ),

    "Extra Trees": ExtraTreesClassifier(
        n_estimators=300,
        min_samples_leaf=2,
        class_weight="balanced",
        random_state=42,
        n_jobs=1
    )
}

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

comparison_results = []

for name, model in models.items():
    pipeline = Pipeline([
        ("preprocessor", preprocessor),
        ("model", model)
    ])

    scores = cross_validate(
        pipeline,
        X_train,
        y_train,
        cv=cv,
        scoring={
            "roc_auc": "roc_auc",
            "f1": "f1",
            "recall": "recall",
            "precision": "precision"
        },
        n_jobs=-1
    )

    comparison_results.append({
        "Model": name,
        "ROC-AUC": scores["test_roc_auc"].mean(),
        "F1-Score": scores["test_f1"].mean(),
        "Attrition Recall": scores["test_recall"].mean(),
        "Precision": scores["test_precision"].mean()
    })

comparison_df = pd.DataFrame(comparison_results)

display(
    comparison_df.sort_values(
        "ROC-AUC",
        ascending=False
    ).round(4)
)
```

```python
from sklearn.model_selection import RandomizedSearchCV
from scipy.stats import randint

# Build the complete pipeline
rf_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("model", RandomForestClassifier(
        random_state=42,
        n_jobs=1
    ))
])

# Define the parameter search space
param_distributions = {
    "model__n_estimators": [200, 300, 500, 700],
    "model__max_depth": [None, 10, 15, 20, 30],
    "model__min_samples_split": randint(2, 15),
    "model__min_samples_leaf": randint(1, 8),
    "model__max_features": ["sqrt", "log2", 0.7],
    "model__class_weight": [
        "balanced",
        "balanced_subsample",
        None
    ]
}

# Search using training data only
random_search = RandomizedSearchCV(
    estimator=rf_pipeline,
    param_distributions=param_distributions,
    n_iter=30,
    scoring="roc_auc",
    cv=StratifiedKFold(
        n_splits=5,
        shuffle=True,
        random_state=42
    ),
    random_state=42,
    n_jobs=-1,
    verbose=1,
    refit=True
)

random_search.fit(X_train, y_train)

print("Best Cross-Validation ROC-AUC:",
      round(random_search.best_score_, 4))

print("\nBest Parameters:")
print(random_search.best_params_)
```

```python
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    roc_auc_score,
    ConfusionMatrixDisplay
)

best_model = random_search.best_estimator_

y_pred_tuned = best_model.predict(X_test)
y_prob_tuned = best_model.predict_proba(X_test)[:, 1]

print("Test Accuracy:",
      round(accuracy_score(y_test, y_pred_tuned), 4))

print("Test ROC-AUC:",
      round(roc_auc_score(y_test, y_prob_tuned), 4))

print("\nClassification Report:")
print(classification_report(
    y_test,
    y_pred_tuned,
    target_names=["Existing Customer", "Attrited Customer"]
))

ConfusionMatrixDisplay.from_predictions(
    y_test,
    y_pred_tuned,
    display_labels=["Existing", "Attrited"],
    cmap="Blues"
)

plt.title("Tuned Random Forest — Test Confusion Matrix")
plt.tight_layout()
plt.show()
```

## Model Explainability — Feature Importance

```python
from sklearn.inspection import permutation_importance

importance_result = permutation_importance(
    best_model,
    X_test,
    y_test,
    scoring="roc_auc",
    n_repeats=5,
    random_state=42,
    n_jobs=-1
)

importance_df = pd.DataFrame({
    "Feature": X_test.columns,
    "Importance": importance_result.importances_mean,
    "Std": importance_result.importances_std
}).sort_values("Importance", ascending=False)

display(importance_df.head(15).round(4))
```

```python
top_features = importance_df.head(10).sort_values("Importance")

plt.figure(figsize=(10, 6))

plt.barh(
    top_features["Feature"],
    top_features["Importance"],
    xerr=top_features["Std"]
)

plt.title("Top 10 Features for Customer Attrition Prediction")
plt.xlabel("Decrease in ROC-AUC When Shuffled")
plt.tight_layout()
plt.show()
```

```python
from sklearn.base import clone
from sklearn.model_selection import cross_val_predict
from sklearn.metrics import precision_recall_curve

threshold_model = clone(best_model)

oof_probabilities = cross_val_predict(
    threshold_model,
    X_train,
    y_train,
    cv=StratifiedKFold(
        n_splits=5,
        shuffle=True,
        random_state=42
    ),
    method="predict_proba",
    n_jobs=-1
)[:, 1]

precision, recall, thresholds = precision_recall_curve(
    y_train,
    oof_probabilities
)

print("Out-of-fold predictions generated:", len(oof_probabilities))
```

```python
f1_scores = (
    2 * precision[:-1] * recall[:-1]
    / (precision[:-1] + recall[:-1] + 1e-12)
)

best_idx = f1_scores.argmax()
best_threshold = thresholds[best_idx]

print("Selected threshold:", round(best_threshold, 4))
print("Cross-validated precision:", round(precision[best_idx], 4))
print("Cross-validated recall:", round(recall[best_idx], 4))
print("Cross-validated F1-score:", round(f1_scores[best_idx], 4))
```

```python
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    roc_auc_score
)

# Use the already-fitted best model
test_probabilities = best_model.predict_proba(X_test)[:, 1]

y_pred_threshold = (
    test_probabilities >= best_threshold
).astype(int)

print("Selected Threshold:", round(best_threshold, 4))
print("Test Accuracy:", round(
    accuracy_score(y_test, y_pred_threshold), 4
))
print("Test ROC-AUC:", round(
    roc_auc_score(y_test, test_probabilities), 4
))

print("\nClassification Report:")
print(classification_report(
    y_test,
    y_pred_threshold,
    target_names=["Existing Customer", "Attrited Customer"]
))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_threshold))
```
